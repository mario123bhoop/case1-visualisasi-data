```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Daftar Produk & Kategori sesuai spesifikasi
produk_kategori = {
    'Laptop': 'Elektronik',
    'Smartphone': 'Elektronik',
    'Tablet': 'Elektronik',
    'Buku Tulis': 'Pendidikan',
    'Novel': 'Pendidikan',
    'Kamus Bahasa': 'Pendidikan',
    'Meja Belajar': 'Furnitur',
    'Kursi Kantor': 'Furnitur'
}

harga_ref = {
    'Laptop': 8500000,
    'Smartphone': 4500000,
    'Tablet': 3000000,
    'Buku Tulis': 25000,
    'Novel': 75000,
    'Kamus Bahasa': 120000,
    'Meja Belajar': 650000,
    'Kursi Kantor': 450000
}

n_samples = 1000
dates = pd.date_range(start='2024-01-01', end='2024-06-30', freq='D')

data = []
for _ in range(n_samples):
    produk = np.random.choice(list(produk_kategori.keys()))
    kategori = produk_kategori[produk]
    harga = harga_ref[produk]
    
    # Diskrit Rasio (Jumlah Terjual)
    jumlah_terjual = np.random.randint(1, 151) if kategori == 'Pendidikan' else np.random.randint(1, 51)
    
    # Waktu Interval
    tanggal = np.random.choice(dates)
    
    data.append({
        'Tanggal': tanggal,
        'Produk': produk,
        'Kategori': kategori,
        'Harga (Rp)': harga,
        'Jumlah Terjual': jumlah_terjual,
        'Total Pendapatan (Rp)': harga * jumlah_terjual
    })

df = pd.DataFrame(data)
df = df.sort_values('Tanggal').reset_index(drop=True)

# Simpan ke CSV
df.to_csv('dataset_penjualan.csv', index=False)
print("✅ SUCCESS: 1.000 Data Sintetis Operasional Penjualan Berhasil Dibuat!")
```
